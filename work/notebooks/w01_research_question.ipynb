*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which pages an editor should review or refresh first. Actor: a content editor with limited weekly review capacity. Cost of a wrong call: reviewing a page that wasn't actually declining wastes editor time; missing a page that was actually declining means a real traffic loss goes unaddressed.
